Vérifie que l'environnement (uv + PyTorch + scikit-learn) fonctionne en entraînant un modèle jouet.

> Pas de GPU local détecté sur cette machine (`torch.cuda.is_available() == False`). Pour tester sur GPU, utiliser Google Colab ou Kaggle Notebooks (voir README de ce dossier).

```python
import torch
import torch.nn as nn

print("PyTorch:", torch.__version__)
print("CUDA disponible:", torch.cuda.is_available())
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Device utilisé:", device)
```

```python
# Modèle jouet : régression sur des données synthétiques (y = 3x + 2 + bruit)
torch.manual_seed(0)

X = torch.linspace(-5, 5, 200).unsqueeze(1)
y = 3 * X + 2 + torch.randn_like(X) * 0.5

X, y = X.to(device), y.to(device)

model = nn.Linear(1, 1).to(device)
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)
loss_fn = nn.MSELoss()

for epoch in range(200):
    optimizer.zero_grad()
    pred = model(X)
    loss = loss_fn(pred, y)
    loss.backward()
    optimizer.step()
    if epoch % 50 == 0:
        print(f"epoch {epoch:3d} | loss {loss.item():.4f}")

w, b = model.weight.item(), model.bias.item()
print(f"\nAppris : y = {w:.2f}x + {b:.2f} (cible : y = 3x + 2)")
```

Si le poids appris est proche de `3` et le biais proche de `2`, l'environnement est fonctionnel. ✅
